## Scatter plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px        # -> for dataset import
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', marker='o', color='red', s=56, label='Hello')
plt.legend()     # if using hue no need of legend , hue automatically creates legend
```

```python
# scatter plot -> axes level function
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='time', size='size')
```

```python
# relplot -> figure level -> generally square shape
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex', style='time', size='size')
```

## Line plot

```python
gap = px.data.gapminder()
gap
```

```python
temp_df = gap[gap['country']=='India']
temp_df
```

```python
# axes level function
sns.lineplot(data=temp_df, x='year', y='lifeExp', color='red', linestyle=':', marker='d')
```

```python
# using relplot
sns.relplot(data=temp_df, x='year', y='lifeExp', kind='line', marker='o')
```

```python
temp_df = gap[gap['country'].isin(['India', 'Brazil', 'Germany'])]
temp_df
```

```python
sns.relplot(kind='line', data=temp_df, x='year', y='lifeExp', hue='country', style='continent', size='continent')
# size='continent'    -> represent thickness of line
```

```python
sns.lineplot(data=temp_df, x='year', y='lifeExp', hue='country')
plt.grid()
```

```python
# facet plot      ->by facet plot -> on a particular categorical(mostly) column k ke har unique value pr multiple plots banaa skte h
# -> only & only work with figure level function -> work with relplot
# it will not wok with scatterplot and lineplot
sns.relplot(kind='scatter', data=tips, x='total_bill', y='tip', col='sex')
```

```python
# facet plot      ->with facer plot -> on a particular categorical column k ooper multiple plots banaa skte h
sns.relplot(kind='line', data=tips, x='total_bill', y='tip', row='smoker', col='sex')
```

```python
sns.relplot(kind='scatter', data=tips, x='total_bill', y='tip', row='sex', col='day')
```

```python
gap
sns.relplot(data=gap, x='lifeExp', y='gdpPercap', kind='scatter', col='year', col_wrap=4)
```

## Histplot

```python
# figure level = displot
# axes level = histplot, kdeplot, rugplot
```

```python
tips
```

```python
plt.figure(figsize=(12,8))
sns.histplot(data=tips, x='total_bill')
```

```python

sns.displot(data=tips, x='total_bill', kind='hist', height=10, aspect=2)
```

```python
# bins parameter 
sns.displot(data=tips, x='total_bill', kind='hist', bins=20)
```

```python
# It’s also possible to visualize the distribution of a categorical variable using the logic of a histogram. 
# Discrete bins are automatically set for categorical variables

tips
```

```python
sns.displot(data=tips, x='day', kind='hist')

# This is not technically a histogram  -> can be said to count plot
# Discrete bins are automatically set for categorical variables
```

```python
# hue parameter
sns.displot(data=tips, x='tip', kind='hist', hue='sex')
```

```python
# element -> step
sns.displot(data=tips, x='tip', kind='hist', hue='sex', element='step')
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic, x='age', kind='hist', hue='sex', element='step')
```

```python
# faceting using row and col

sns.displot(data=tips, x='tip', kind='hist', col='sex')
```

## kdeplot

```python
# KDE = kernal density estimation
# Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel(ek function ko use krke), producing a continuous density estimate

tips
sns.kdeplot(data=tips, x='total_bill')
```

```python
sns.displot(kind='kde', data=tips, x='total_bill', color='black', fill=True)
```

```python
# hue -> fill
sns.displot(kind='kde', data=tips, x='total_bill', hue='sex', fill=True)
```

```python
tips
```

## Rugplot

```python
# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.

sns.kdeplot(data=tips, x='total_bill')
sns.rugplot(data=tips, x='total_bill')
```

```python
# Bivariate histogram
# A bivariate histogram bins the data within rectangles that tile the plot 
# and then shows the count of observations within each rectangle with the fill color

# sns.histplot(data=tips, x='total_bill', y='tip', hue='sex')
sns.displot(data=tips, x='total_bill', y='tip', kind='hist')
```

```python
# Bivariate Kdeplot
# a bivariate KDE plot smoothes the (x, y) observations with a 2D Gaussian

sns.kdeplot(data=tips, x='total_bill', y='tip')

# how to interpret this graph -> jahan lines ek-dusre ke close h , vahan density zyada h -> vahan zyada customers
                                # agr lines dooor h , vahan kam customers hain
```

## 3. Matrix Plot
#### There is  no figure level function for Matrix plot, both Heat and Cluster are axes level functions
#### i. Heatmap
#### ii. Clustermap

### 1. Heatmap

```python
# Heatmap
# axes level function

# Plot rectangular data as a color-encoded matrix
```

```python
gap
```

```python
import pandas as pd
plt.figure(figsize=(15,10))

grid = pd.pivot_table(gap, index='country', columns='year', values='lifeExp')
grid

sns.heatmap(grid)
```

```python
plt.figure(figsize=(14,10))
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country', columns='year', values='lifeExp')

sns.heatmap(temp_df, annot=True, linewidth=0.5, cmap='viridis')
```

### 2. Clustermap

```python
# is basically similar to heatmap , ye internally data k ooper clustring krta h
# clustring ka mtlb similar columns ko ek-saath kr deta h or , dissimilar columns ko alg kr deta h

# Plot a matrix datset as a hierarchically-clustered heatmap
```

```python
iris = px.data.iris()
iris
```

```python
plt.figure(figsize=(14,10))
sns.clustermap(iris.iloc[:, [0,1,2,3]])
```
