## Strip plot

```python
# Strip plot
# axes level function
tips
sns.stripplot(data=tips, x='day', y='total_bill')
# sns.stripplot(data=tips, x='day', y='total_bill', jitter=False)
```

## Categorical Distribution Plots

```python
# univariate - single particular column ka distribution (aisa kam hota h ki ek numerical variable ki distribution)
# MOSTLY -> Seaborn ke boxplot() function ko sabse zyada categorical vs numerical comparison ke liye use kiya jata hai, jo bivariate analysis hai.
```

### 1. Boxplot
#### A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.



```python
sns.boxplot(data=tips,x='day',y='total_bill')
# sns.boxplot(data=tips,x='total_bill')
```

```python
# uisng catplot
sns.catplot(data=tips,x='day',y='total_bill', kind='box', hue='sex')
```

```python
sns.boxplot(data=tips,x='day',y='total_bill', hue='smoker')
```

```python
tips
```

```python
# single boxplot -> numeical column
sns.boxplot(data=tips,x='total_bill')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='violin',hue='sex')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='violin',hue='sex', split=True)
```

## Categorical Estimate Plot -> for central tendency

```python
# for central tendency -> like estimate some value, like mean or mean or count or some aggrate value
```

### 1. Bar plot

```python
sns.barplot(data=tips, x='sex', y='total_bill')
```

```python
import numpy as np
sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker', estimator='mean')
```

```python
sns.catplot(data=tips, x='sex', y='total_bill', errorbar=None, kind='bar', estimator=np.min)
```

```python
# When there are multiple observations in each category, it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars
```

### 2. Point plot

```python
sns.pointplot(data=tips, x='sex', y='total_bill', hue='smoker', estimator='mean')
```

sns.catplot(data=tips, x='sex', y='total_bill', kind='point')    # basically ye dono k beech k difference ka estimate de rha %%html

```python
sns.pointplot(data=tips, x='sex', y='total_bill', errorbar=None)    # basically ye dono k beech k difference ka estimate de rha %%html
```

### 3. Count plot

```python
# A special case for the bar plot is when you want to show the number of observations in each category rather than computing a statistic for a second variable. This is similar to a histogram over a categorical, rather than quantitative, variable
```

```python
sns.countplot(data=tips, x='sex')
```

```python
sns.catplot(data=tips, x='sex', hue='smoker', kind='count')
```

```python
sns.countplot(data=tips, x='sex', hue='day')
```

```python
# faceting using catplot
sns.catplot(data=tips, x='sex',y='total_bill',kind='box', col='smoker', row='time')
```

# Regression plot (Regplot)

### regplot
### lmplot
#### In the simplest invocation, both functions draw a scatterplot of two variables, x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression.

```python
tips
```

```python
# axes level function
# hue parameter is not available
sns.regplot(data=tips, x='total_bill', y='tip')
```

```python
# figure level 
# hue is available
sns.lmplot(data=tips, x='total_bill', y='tip', hue='sex')

# scatterplot() ↔ relplot()
# regplot() ↔ lmplot()
```

```python
# residplot (residual plot)

sns.residplot(data=tips, x='total_bill', y='tip')
```
